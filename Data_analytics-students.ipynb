# Gender Distribution


```python
ax=sns.countplot(data=df, x="Gender")
ax.bar_label(ax.containers[0])
plt.title("Count Of Genders")
plt.show()
```

### #From the above chart, we've analysed that the number of females is more the the number of males

```python
gb = df.groupby("ParentEduc").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb)
```

```python
sns.heatmap(gb, annot=True)
plt.title("Relationship between Parent's Education and Students' Scores")
plt.show()
```

## #From the above chart, we've concluded that the education of parents have a good impact on the scores of their kids.

```python
gb1 = df.groupby("ParentMaritalStatus").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb1)
```

```python
sns.heatmap(gb1, annot=True)
plt.title("Relationship between Parent's Marital Status and Students' Scores")
plt.show()
```

## #From the above chart, we've analyzed that the marital status of parents have a negligible impact on the scores of students.

```python
sns.boxplot(x="MathScore", data=df)
plt.show()
sns.boxplot(x="ReadingScore", data=df)
plt.show()
sns.boxplot(x="WritingScore", data=df)
plt.show()

```

## #From the above box plots, we can conclude that Maths is comparatively difficult to score.

# Distribution of Ethnic Groups:

```python
groupA = df.loc[(df['EthnicGroup']=="group A")].count()
groupB = df.loc[(df['EthnicGroup']=="group B")].count()
groupC = df.loc[(df['EthnicGroup']=="group C")].count()
groupD = df.loc[(df['EthnicGroup']=="group D")].count()
groupE = df.loc[(df['EthnicGroup']=="group E")].count()

l=["group A", "group B", "group C", "group D", "group E"]
mlist = [groupA["EthnicGroup"],groupB["EthnicGroup"],groupC["EthnicGroup"],groupD["EthnicGroup"],groupE["EthnicGroup"]]

plt.pie(mlist, labels=l,autopct='%1.2f%%')
plt.title("Distribution of Eth")
plt.show()
```
